```python
import pandas as pd, matplotlib.pyplot as plt
from app.database import engine
df = pd.read_sql('select * from jobs', engine)
df.shape
```

```python
df.role_family.value_counts().plot.barh(title='Postings by role family'); plt.show()
```

```python
df.groupby('role_family')[['salary_min','salary_max']].median().sort_values('salary_max').plot.barh(title='Median salary range (INR/yr)'); plt.show()
```

```python
from app.models import Skill, job_skills
q = '''select s.name, count(*) n from job_skills js join skills s on s.id=js.skill_id group by s.name order by n desc limit 15'''
pd.read_sql(q, engine).set_index('name').n.plot.barh(title='Top skills'); plt.show()
```
